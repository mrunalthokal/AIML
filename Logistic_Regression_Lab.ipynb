```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
col_names = ['pregnant','glucose','bp','skin','insuline','bmi','pedigree','age','label']
```

```python
df = pd.read_csv("diabetes_dataset.csv",names=col_names)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df['age'] = pd.to_numeric(df['age'],errors='coerce')
```

```python
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_encoded = df.apply(le.fit_transform)
```

```python
df_encoded.head()
```

```python
bins = [0,18,35,60,100]
labels = ['Child','Young','Adult','Senior']

df['Age Group'] = pd.cut(df['age'],bins=bins,labels=labels)
```

```python
df.head()
```

```python
df['glucose'] = pd.to_numeric(df['glucose'],errors='coerce')
df['label'] = pd.to_numeric(df['label'],errors='coerce')
df = df.dropna(subset=['glucose','label'])
x=df[['glucose']]
y=df[['label']]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_train)
```

```python
predicted_probs = model.predict_proba(x_train)[:,1]
```

```python
glucose_range = np.linspace(df['glucose'].min(),df['glucose'].max(),300).reshape(-1,1)
glucose_range_df = pd.DataFrame(glucose_range,columns=['glucose'])
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='glucose',y='label',color='blue',data=df,alpha=0.6,label='data')
plt.plot(glucose_range_df,predicted_probs,color='pink',linewidth=2,label='Logictic Regression')
plt.title("Logistic Regression Curve")
plt.xlabel("Glucose")
plt.ylabel("Label")
plt.grid(True)
plt.legend()
plt.show()
```
